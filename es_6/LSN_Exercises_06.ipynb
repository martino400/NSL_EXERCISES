In Statistical Mechanics the one-dimensional (1D) Ising Model is exactly resoluble. The Hamiltonian for a system of N spins with nearest neighbors interaction is given by (in the following I will assume units such that $\mu_B=1$ and $k_B=1$):

$$
H = 
-J\sum_{i=1}^N s_i s_{i+1}
-\frac{h}{2}\sum_{i=1}^N (s_i + s_{i+1}) 
$$

where $\{ s_i \}_{i=1,N}$ are spin variables taking values in $\{-1;1\}$, $J>0$ (ferromagnetic model),and with the assumption $s_{N+1} = s_{1}$.

The partition function for $N$ spins, $Z$, can be exactly computed and it turns out to be:

$$
Z=\sum_{\{ s_i \}} \exp \left( -\beta H\right) = \lambda_1^N + \lambda_2^N
$$

where
$\lambda_{1,2} = 
e^{\beta J} 
\cosh (\beta h) \pm \left[ e^{2\beta J} \cosh^2 (\beta h) - 2\sinh (2\beta J) \right]^{1/2}$, 
and where $\beta=\frac{1}{k_b T} = \frac{1}{T}$.
Once the partition function has been computed, the Helmholtz free energy is readly obtained as $A = -k_B T \ln Z$. All the thermodynamic quantities of interest can be deduced from $A$ or, equivalently, from $Z$:

- the **Internal Energy** per spin (with $h=0$):

$$
\frac{U(N,T)}{N} = \frac{\langle H \rangle}{N}
= \frac{1}{N} \sum_{\{ s_i \}} H \frac{\exp \left( -\beta H\right)}{Z}
= -\frac{1}{N}\frac{\partial \ln Z}{\partial \beta}
\underset{h=0}= -J \tanh (\beta J) \Biggl( \frac{1 + \tanh^{N-2} (\beta J)}{1+\tanh^N (\beta J)} \Biggl)
$$

- the **Heat Capacity** (with $h=0$):

$$
C(N,T) = \frac{\partial U(N,T)}{\partial T}
= \frac{\partial \beta}{\partial T}\frac{\partial U}{\partial \beta}
= -k_B\beta^2 \frac{\partial}{\partial \beta} \left[ \sum_{\{ s_i \}} H \frac{\exp \left( -\beta H\right)}{Z} \right] = \\
= -k_B\beta^2 \left[ -\sum_{\{ s_i \}} H^2 \frac{\exp \left( -\beta H\right)}{Z} + \sum_{\{ s_i \}} H \exp \left( -\beta H\right)\frac{\partial Z^{-1}}{\partial \beta} \right] = \\
= k_B\beta^2 \left\{ \sum_{\{ s_i \}} H^2 \frac{\exp \left( -\beta H\right)}{Z} + \left[ \sum_{\{ s_i \}} H \frac{\exp \left( -\beta H\right)}{Z}\right]\frac{1}{Z}\frac{\partial Z}{\partial \beta} \right\} =\\
= k_B\beta^2 \left\{ \sum_{\{ s_i \}} H^2 \frac{\exp \left( -\beta H\right)}{Z} - \left[ \sum_{\{ s_i \}} H \frac{\exp \left( -\beta H\right)}{Z}\right]^2 \right\} =
k_B\beta^2 \left(\langle H^2 \rangle -\langle H \rangle^2 \right)
$$

$$
\frac{C(N,T)}{N} = \frac{1}{N}\frac{\partial U(N,T)}{\partial T} =\\
\underset{h=0}= 
k_B (\beta J)^2 \left\{ 1+\tanh^N (\beta J)+(N-1)\tanh^2 (\beta J)+(N-1)\frac{\tanh^{N-2} (\beta J)}{1+\tanh^N (\beta J)}
- N \left[ \frac{\tanh (\beta J) + \tanh^{N-1} (\beta J)}{1+\tanh^N (\beta J)} \right]^2 \right\}
$$

- The **magnetization** (with $h \not= 0$):

Note that with $h=0$ the magnetization $M(N,T,h=0)=\langle \sum_{i=1,N}s_i \rangle =0$ because in 1D we cannot have a (ferromagnetic) phase transition in absence of an external magnetic field.

$$
M(N,T,h) = -\frac{\partial A}{\partial h} = k_B T \frac{\partial \ln Z}{\partial h} =
\frac{1}{\beta} \frac{1}{Z}\frac{\partial}{\partial h}\sum_{\{ s_i \}} \exp \left( -\beta H\right) = \sum_{\{ s_i \}} \left[ \sum_{i=1,N}s_i\right] \frac{\exp \left( -\beta H\right)}{Z} = \left\langle \sum_{i=1,N}s_i \right\rangle
$$

The exact formula from $M(N,T,h) = k_B T \dfrac{\partial \ln Z}{\partial h}$ is:
$$
\frac{M(N,T,h)}{N}=\frac{e^{\beta J}\; \sinh(\beta h)}{Z} \; \Bigl[\lambda_1^{N-1}\cdot \Bigl(1+\frac{e^{\beta J}\; \cosh(\beta h)}{\sqrt{e^{2\beta J} \cosh^2 (\beta h) - 2\sinh (2\beta J)}}\Bigr) + \lambda_2^{N-1}\cdot \Bigl(1-\frac{e^{\beta J}\; \cosh(\beta h)}{\sqrt{e^{2\beta J} \cosh^2 (\beta h) - 2\sinh (2\beta J)}}\Bigr)\Bigr]
$$

- The **magnetic susceptibility** (with $h=0$):

$$
\chi(N,T) = \frac{\partial M}{\partial h} = 
\frac{\partial}{\partial h}\sum_{\{ s_i \}} \left[ \sum_{i=1,N}s_i\right] \frac{\exp \left( -\beta H\right)}{Z} =\\
= \beta\sum_{\{ s_i \}} \left[ \sum_{i=1,N}s_i\right]^2 \frac{\exp \left( -\beta H\right)}{Z}
-\sum_{\{ s_i \}} \left[ \sum_{i=1,N}s_i\right] \frac{\exp \left( -\beta H\right)}{Z}
\frac{1}{Z}\frac{\partial Z}{\partial h} = 
\beta \left[ \left\langle \left( \sum_{i=1,N}s_i \right)^2 \right\rangle - \left\langle \sum_{i=1,N}s_i \right\rangle^2 \right] = \\
\underset{h=0}= \beta \left[ \left\langle \left( \sum_{i=1,N}s_i \right)^2 \right\rangle \right] = \beta e^{2\beta J} \Biggl( \frac{1-\tanh^N (\beta J)}{1+\tanh^N (\beta J)} \Biggl)
$$

### **Exercise 06.1**
To execute this exercise it was necessary to use two sampling methods: $\textbf{Metropolis and Gibbs}$ methods. These are two sampling methods used to sample element of the configuration space sampled from the probability distribution function $\rho(\mathbf{q}, \mathbf{p})$. The mathematical field behind these these methods are the $\textit{Stochastic Markov processes}$. 
With Metropolis technique (or $M(RT^2)$) we already know what the desiderd $p(\mathbf{x})$ (probability distribution function) is. Where $\mathbf{x}$ is a vector of the configuration space, in general a many-dimensional vector (e.g: the position of all the particles or the spin variables of all the atoms of a ferromagnetic material). 
#### **Metropolis sampling**
The $M(RT^2)$ algorithm is a particular type of Markov process that implements a transition probability $T(y|x)$, which is the probability of going from the state $x$ to $y$ so that, when repeated for $n_{steps}\to\infty$, it yields the distribution $p(\mathbf{x})$. Defining $d_n(x)$ as the probability distribution function associated Markov step $n$. It is possible to write the condition above as follows:
\begin{equation}
    \lim_{n\to\infty} d_n(x) = p(x)
    \tag{1}
\end{equation}
Furthermore it is necessay to make another hypothesis which is that the random walk in the phase space must be $\textbf{stationary}$:
$$
\begin{aligned}
    p_1 (x, t+\Delta t) &= p_1(x,t) = p(x) \tag{2} \\
    p_{1|1}(x_2,t_2|x_1, t_1) &= p_t(x_2,x_1) = K(x_2, x_1)
\end{aligned}
$$
We remember that $p_{1|1}$ is probability of transitioning from the step $1$ and performing $1$ step and we defined $t = t_2 - t_1$, which we consider equal for all timesteps and therefore constant throughout all the process. This last equation physically means that the transition probability isn't dependent on the time variable but only on the state of the system. From the conservation of probability we can write the second equation that characterizes stochastic Markov processes:
\begin{equation}
    p(x) = \int K(x|y)p(y)dy
    \tag{3}
\end{equation}
Which means that the probability function $p(x)$ is invariant with respect to the transition probability, which intuitively understandable since, having asymptotically reached the probability density function, it wouldn't make sense for the $p(x)$ to vary according to $K$.
After some further definitions, theorems and calculations we find that:
\begin{equation}
    K(y|x) = T(y|x) A(y|x) = T(y|x) \min[1; \frac{p(y)}{p(x)}]
    \tag{4}
\end{equation}
In the $\textbf{1D Ising Model}$ we find that the acceptance ratio $A(E_f|E_i) = e^{-\beta \Delta E}$, where to sample we use the energy function of the system. Suppose we are interested in flipping the spin $i$
\begin{equation}
    \Delta E =  2 s_i (J(s_{i+1} + s_{i-1}) + h)
    \tag{5}
\end{equation}



#### **Gibbs sampling**
The Gibbs sampling algorithm works by keeping all variables of the system fixed (in this case $N$ spins except one) and changing only one. Therefore, the acceptance rate is 1 for the Gibbs sampling, all moves are accepted. The transition probability for this process can be written as follows:
\begin{equation}
    T_k(\mathbf{x}|\mathbf{y}) = p_{1|n-1}(x_k|\{y_i : i \neq k\}) \prod_{i \neq k} \delta(x_i - y_i)
    \tag{6}
\end{equation}
Which means that all variables $y_i$ remain the same ($x_i = y_i$) except for the $y_k$ variable which is transitions into a new variable $y_k$. The conditional probability distribution for $x_1$ given $x_2$ is given by the ratio $p_{1|1}(x_1|x_2)=p(x_1,x_2)/p_1(x_2)$ where

$$p_1(x_2) = \int dx_1 p(x_1, x_2)$$

is the marginal distribution for the $x_2$ variable and that $p_{1|1}(x_1|x_2)$ is a probability distribution once $x_2$ has been fixed.

In the $\textbf{1D Ising Model}$ directly from equation (6) we find that:
\begin{equation}
    p_{1|1}(s_k'| s_i \{i \neq k \}) = \frac{p(s_1,...,s_k',...,s_n)}{p(s_1,...,s_k =1,...,s_n) + p(s_1,...,s_k =1,...,s_n)} = \frac{1}{1+e^{- \beta \Delta E}}
    \tag{7}
\end{equation}
Where in the last equation I made use of the fact that $s_k = \pm 1$. Therefore it doesn't make sense to speak of acceptance for the Gibbs sampling method since for this exercise based on the value of found the random number the spin is either set to 1 or -1.

````python
import numpy as np
import matplotlib.pyplot as plt

# 1. Set a clean, professional visual style
plt.style.use('seaborn-v0_8-whitegrid')

# 2. Data Loading
# Dati Gibbs
chi_gibbs = np.loadtxt('DATA1/file0_gibbs.dat')
cv_gibbs = np.loadtxt('DATA1/file2_gibbs.dat')
internal_gibbs = np.loadtxt('DATA1/file1_gibbs.dat')
magn_gibbs = np.loadtxt('DATA1/file3_gibbs.dat')

# Dati Metropolis
chi_metro = np.loadtxt('DATA1/file0_metro.dat')
cv_metro = np.loadtxt('DATA1/file2_metro.dat')
internal_metro = np.loadtxt('DATA1/file1_metro.dat')
magn_metro = np.loadtxt('DATA1/file3_metro.dat')

# 3. Create figure with 4 subplots (2x2 grid)
fig, axs = plt.subplots(2, 2, figsize=(14, 10))

# Palette colori per distinguere Gibbs e Metropolis
color_gibbs = '#E74C3C'  # Rosso per Gibbs
color_metro = '#2E86C1'  # Blu per Metropolis

# Helper function per plottare i punti collegati da linee
def plots(ax, data, color, label, title, x_label=None, marker='o'):
    # x = col 0, y = col 4 (la quinta colonna degli errori)
    # linestyle='-' collega i punti con una linea continua
    ax.plot(data[:, 0], data[:, 4], 
            color=color, marker=marker, linestyle='-', linewidth=1.2, markersize=4, 
            markeredgecolor='black', alpha=0.8, label=label)
    
    ax.set_title(title, fontsize=14, fontweight='bold', pad=10)
    ax.set_ylabel('Value', fontsize=11, fontweight='bold')
    if x_label:
        ax.set_xlabel(x_label, fontsize=11, fontweight='bold')

# ==========================================
# [0, 0] Top Left: INTERNAL ENERGY
# ==========================================
plots(axs[0, 0], internal_gibbs, color_gibbs, 'Gibbs', 'Error of internal Energy (U)', marker='o')
plots(axs[0, 0], internal_metro, color_metro, 'Metropolis', 'Error of internal Energy (U)', marker='s')

# ==========================================
# [0, 1] Top Right: SPECIFIC HEAT
# ==========================================
plots(axs[0, 1], cv_gibbs, color_gibbs, 'Gibbs', 'Error of Specific Heat ($C_v$)', marker='o')
plots(axs[0, 1], cv_metro, color_metro, 'Metropolis', 'Error of Specific Heat ($C_v$)', marker='s')

# ==========================================
# [1, 0] Bottom Left: MAGNETIZATION
# ==========================================
plots(axs[1, 0], magn_gibbs, color_gibbs, 'Gibbs', 'Error of Magnetization (M)', x_label='Temperature (T)', marker='o')
plots(axs[1, 0], magn_metro, color_metro, 'Metropolis', 'Error of Magnetization (M)', x_label='Temperature (T)', marker='s')

# ==========================================
# [1, 1] Bottom Right: SUSCEPTIBILITY
# ==========================================
plots(axs[1, 1], chi_gibbs, color_gibbs, 'Gibbs', 'Error of Susceptibility ($\chi$)', x_label='Temperature (T)', marker='o')
plots(axs[1, 1], chi_metro, color_metro, 'Metropolis', 'Error of Susceptibility ($\chi$)', x_label='Temperature (T)', marker='s')

# ==========================================
# Common Customization for all subplots
# ==========================================
for ax in axs.flat:
    # Remove top and right spines for a modern look
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    # Add legend with a slight shadow
    ax.legend(frameon=True, shadow=True, fontsize=10, loc='best')

# Global Title
fig.suptitle('Gibbs vs Metropolis Error Results', 
             fontsize=18, fontweight='bold', y=1.02)

# 4. Final layout optimization
plt.tight_layout()

# Save command
plt.savefig('Images/GibbsMetropolisError.png', dpi=300, bbox_inches='tight')

plt.show()
````

- I did the simulation for a range of values of temperature from T = 2.0 to T = 0.5 by decreasing the temperature of 0.1 at every cycle. Furthermore, for each point of the simulation I extrapolated the last value of the data-blocking done for each temperature.

- We can the that the data is compatible with the theoretical values. In certain points we can see that the data used with the Metropolis sampling isn't compatible, this occurs due to the fact that for that temperature I should have used more steps. 

- Another thing that it is possible to observe from the data obtained from both algorthsm: the errors on the data blocking become greater as the temperature decreases (these values in `DATA1` are values of a simulation executed for the same number of total steps for all temperature).
