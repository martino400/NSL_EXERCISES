### Exercise 02.1
- Compute the following 1D integral via Monte Carlo 
$$I = \int_0^1 \frac{\pi}{2}\cos(\pi x/2) dx = 1$$

1. sampling a uniform distribution in $[0,1]$
2. using importance sampling (i.e. sampling a non-uniform probability in $[0,1]$)

Per fare il sampling ho sviluppato con taylor in 1 al primo ordine 


```python
import matplotlib
import matplotlib.pyplot as plt
import numpy as np

#esercizio 2.1.1 e 2.1.2


# Caricamento dati
x, f, error1 = np.loadtxt("esercizio2.1/data/Data2.1.1.dat", usecols=(0,1,2), delimiter='\t', unpack=True)
x2, f2, error2 = np.loadtxt("esercizio2.1/data/Data2.1.2.dat", usecols=(0,1,2), delimiter='\t', unpack=True)

# Stile grafico
plt.style.use('seaborn-v0_8-darkgrid')

plt.figure(figsize=(10,6))

# Grafici con barre d'errore
plt.errorbar(
    x, f, yerr=error1,
    fmt='o-', markersize=4,
    capsize=3, elinewidth=1,
    label='Uniform distribution'
)

plt.errorbar(
    x2, f2, yerr=error2,
    fmt='s-', markersize=4,
    capsize=3, elinewidth=1,
    label='Importance sampling'
)

plt.axhline(
    y=1.0,
    color='red',
    linestyle='--',
    linewidth=1.5,
    label='Expected Integral value'
)

# Etichette
plt.xlabel('Number of blocks', fontsize=13)
plt.ylabel('Integral value', fontsize=13)

# Titolo
plt.title('Confronto: Uniform Sampling vs Importance Sampling', fontsize=15)

# Legenda
plt.legend(fontsize=11)

# Migliora margini
plt.tight_layout()

plt.show()


#da mettere in marcdown il commento e quello che ho fatto

#si vede che con l'importance sampling è diminuita l'incertezza rispetto alla distribuzione uniforme
```

### Exercise 02.2
- <span style="color:red">3D Random Walks (RW) on a cubic lattice and in the continuum:</span> Repeat many times (e.g. say $10^4$) the simulation of a random walk in 3D always starting at the origin

1. on a cubic lattice with lattice constant $a=1$; at each discrete time the walker makes a forward or backward step of length equal to $a$ in one of the 3 principal directions of the lattice: $x$, $y$ or $z$
2. in the continuum; at each discrete time the walker makes a step of length equal to $a(=1)$ along a **random direction** obtained by sampling **uniformly** the solid angle: $\theta \in [0,\pi]$ and $\phi \in [0,2\pi]$
<span style="color:red">Show a picture of $\sqrt{\langle |\vec{r}_N|^2 \rangle_{RW}}$ for both RWs</span>, with their statistical uncertainties, as function of the step $i\in [0,10^2]$.

Suggestion: divide your $M$ throws into a reasonable number of blocks, compute $\sqrt{\langle |\vec{r}_N|^2 \rangle_{RW}}$ as function of the step $i\in [0,10^2]$ in each block and use these values (for fixed $i$) to obtain the average value and its statistical uncertainty.

Note that you could try to fit both results with a function like $f(N)=k\sqrt{N}$. Do your results indicate a diffusive behavior?

```python
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
from scipy.optimize import curve_fit

# =========================
# Modello di fit
# =========================
def model(x, a):
    return a * np.sqrt(x)

# =========================
# Funzione per caricare + fittare + plottare
# =========================
def process_and_plot(ax, file_path, title):
    
    if not file_path.exists():
        raise FileNotFoundError(f"File not found: {file_path}")
    
    x, y, err = np.loadtxt(file_path, unpack=True)

    # Fit pesato
    popt, pcov = curve_fit(
        model,
        x,
        y,
        sigma=err,
        absolute_sigma=True
    )

    a_fit = popt[0]
    a_err = np.sqrt(pcov[0, 0])

    print(f"{title} -> a = {a_fit:.5f} ± {a_err:.5f}")

    # Curve lisce
    x_fit = np.linspace(min(x), max(x), 300)
    y_fit = model(x_fit, a_fit)
    y_theory = np.sqrt(x_fit)

    # Plot dati
    ax.errorbar(
        x, y,
        yerr=err,
        fmt='o',
        markersize=4,
        capsize=3,
        label='Simulazione'
    )


#Da sistemare l'errore sul parametro del fit

    # Fit
    ax.plot(
        x_fit, y_fit,
        linewidth=2,
        label=fr'Fit: $a\sqrt{{N}},\ a={a_fit:.3f}$'
    )

    # Teoria
    ax.plot(
        x_fit, y_theory,
        linestyle='--',
        linewidth=2,
        label=r'Teoria: $\sqrt{N}$'
    )

    # Labels
    ax.set_title(title, fontsize=13)
    ax.set_xlabel('N')
    ax.set_ylabel(r'$\sqrt{\langle r^2 \rangle}$')
    ax.legend()
    ax.grid(True)


# =========================
# Figure con subplot
# =========================
plt.style.use('seaborn-v0_8-darkgrid')

fig, axes = plt.subplots(1, 2, figsize=(14, 6), sharey=True)

# File paths
file_cubic = Path("esercizio2.2/data/ValuesRWCubic.dat")
file_cont = Path("esercizio2.2/data/ValuesRWContinuos.dat")

# Plot
process_and_plot(axes[0], file_cubic, "Random Walk 3D - Cubic lattice")
process_and_plot(axes[1], file_cont, "Random Walk 3D - Continuous")


plt.show()
```
