### Exercise 03.1 <span style="color:red"> Plain vanilla option pricing </span>

Black-Scholes <span style="color:blue"> analytic solution </span> for the price at time t<T:
- European Call-option price:
$$ C\left[ S(t),t \right] = S(t) N(d_1) -K e^{-r(T-t)} N(d_2) $$
- European Put-option price:
$$ P\left[ S(t),t \right] = S(t) \left[ N(d_1)-1\right] -K e^{-r(T-t)} \left[ N(d_2)-1\right]$$

where 
$$d_1 = \frac{1}{\sigma\sqrt{T-t}}\left[ \ln \left( \frac{S(t)}{K} \right) + \left( r+\frac{\sigma^2}{2} (T-t) \right)\right]$$

where $d_2 = d_1 - \sigma\sqrt{T-t}$, and where
$$ N(x) = \int_{-\infty}^{x} dz \, e^{-\frac{z^2}{2}} = \frac{1}{2} \left[ 1+ {\rm erf} \left( \frac{x}{\sqrt{2}}\right) \right]$$
is the cumulative of Gaussian $\sim \mathcal{N}(\mu=0,\sigma^2=1)$,
and where
$${\rm erf}(x) = \frac{2}{\sqrt{\pi}}\int_{0}^{x} dz \, e^{-z^2}$$

By using the following parameters: 
- asset price at $t=0$: $S(0)=100$
- delivery time: $T=1$
- strike price: $K=100$
- risk-free interest rate: $r=0.1$
- volatility: $\sigma=0.25$

compute at time $t=0$ **via Monte Carlo** the European call-option price, $C[S(0),0]$, and put-option price, $P[S(0),0]$:

1. <span style="color:red"> By sampling **directly** the final asset price $S(T)$</span> for a $GBM(r,\sigma^2)$
2. <span style="color:red"> By sampling the discretized $GBM(r,\sigma^2)$ path of the asset price</span> 
dividing $[0,T]$ in $10^2$ time intervals: $S(0)$, $S(t_1)$, $S(t_2)$, ... up to $S(t_{100}=T=1)$

<span style="color:red">Make four pictures for the estimation of the European call-option prices, $C[S(0),0]$ (direct and discretized), and put-option prices, $P[S(0),0]$ (direct and discretized), with their uncertainties </span> with a large number of asset prices at time $t=T$, say $M\ge 10^4$, as a function of the number of blocks, $N$</font>. <span style="color:blue"> As usual, in computing statistical uncertainties, use data blocking.</span>

### For comparison, Black-Scholes analytic solution:

````python
import scipy
from scipy.special import erf
import numpy
from numpy import exp
from numpy.lib.scimath import log, sqrt

# data
S0 = 100.
K = 100.
T = 1.
r = 0.1
sigma = 0.25
t=0

def N(x):
    return 0.5 * (1. + erf(x / sqrt(2.)))

def black_scholes(S0, K, T, r, sigma):
    d1 = 1./(sigma * sqrt(T)) * (log(S0 / K) + (r + (sigma**2) / 2.) * T)
    d2 = d1 - sigma * sqrt(T)
    C = S0 * N(d1) - K * exp(-r * T) * N(d2)
    P = S0 *(N(d1) - 1.) - K * exp(-r * T) * (N(d2)-1.)
    return C, P
if __name__ == "__main__":
    call, put = black_scholes(S0, K, T, r, sigma)
    print(" == BLACK-SCHOLES ==")
    print("call: ", call)
    print("put: ", put)
````

### **Exercise 03.1 <span style="color:red"> Plain vanilla option pricing </span>**

Black-Scholes theory assumes that the time evolution of the asset price is stochastic and exhibits geometric Brownian motion (GBM) with constant risk-free interest rate $r$ and volatility $\sigma$.

dividing $[0,T]$ in $10^2$ time intervals: $S(0)$, $S(t_1)$, $S(t_2)$, ... up to $S(t_{100}=T=1)$

- asset price at $t=0$: $S(0)=100$
- delivery time: $T=1$

- risk-free interest rate: $r=0.1$

compute at time $t=0$ **via Monte Carlo** the European call-option price, $C[S(0),0]$, and put-option price, $P[S(0),0]$:

#### **Theoretical framework**
An option is a financial contract between two parties: one has assumes an obligation whilst the other assumes obtains a right. The aim of this exercise is to simulate European option pricing. A **call** option is when the seller (writer) of the option grants the buyer (holder) a right to purchase an underlying (some sort of purchasable item) at a specified price $K$ called strike price at a certain time T in the future. A **put** option on the other hand is when the seller has the right to **sell** the option at a decided price $K$ at a future time T. Therefore, the holder (the party that has the right in the contract) will exploit his right only when profitable. The profit for **call** and **put** are respectively:

$\begin{equation}
 \text{profit} = \max[0, S(T) - K] \text{ for a call;       } \text{profit} = \max[0, K-S(T)] \text{ for a put;}
\end{equation}$

In the case of a call the option becomes profitable once $S(T)$ becomes greater than the delivery price, therefore the holder has the right to buy at a lower price than the market one. The put option becomes profitable $S(T)$ becomes profitable once the market price becomes lower than the delivery price since the holder can exploit his right to sell the asset at the decided price of the contract. 

Black and Scholes tried to determine how much an option should cost based on 2 simple assumptions:
1. The market is infinitely liquid and doesn't exhibit any friction (which means that they neglected all possible bank commisions and taxes and that all transactions happen instaneneously. 
2. The evolution of the asset price over time can be seen as a Brownian motion with **constant** risk-free interest rate and volatility (which measures the strenghts of the price fluctuations).

The assumption of constant volatility proved to be inefficient and far from the realistic evolution of the asset market since the volatility is not constant in reality, it fluctuates with time and periods of high volatility tend to be correlated. This lead to the model being inefficien making the hedge-fund founded by Scholes to lose $4.6 bilion dollars in only 4 months. 


<p style="border:2px; border-style:solid; border-color:#F5F5F5; padding: 1em; background-color:#F5F5F5">
    To perform the exercise it was fundamental to remember that given an asset price $S\sim GBM(\mu,\sigma^2)$, with initial value $S(0)$, then
    $$S(t) = S(0) \exp\left[ \left(\mu-\frac{1}{2}\sigma^2\right)t+\sigma W(t)\right]$$
    where $W(t)\sim\mathcal{N}(0,t)$.
    Moreover, since the increments of $W$ are independent and normally distributed, the simple recursive procedure for sampling discretized values of $S\sim GBM(\mu,\sigma^2)$ at $t_1 < t_2 < … < t_n$ is:
$$S(t_{i+1}) = S(t_i) \exp\left[ \left(\mu-\frac{1}{2}\sigma^2\right)(t_{i+1}-t_i)
    +\sigma Z_{i+1}\sqrt{t_{i+1}-t_i}\right]$$
    with $Z_i \sim \mathcal{N}(0,1)$
</p>


#### **Code structure**
To execute this exercise I created two functions: 
- `void Black_Scholes_Direct(vector<double> &C, vector<double> &P, Random rnd, int M)` which executes $M$ times a direct Monte-Carlo calculation of the call and put profit values at the delivery time $T$ using the direct evolution of $S(T)$ written above.
- `void Black_Scholes_Step_by_Step(vector<double> &C, vector<double> &P, Random rnd, int M)` which executes $M$ times a discretized sampling of $S(t)$ using the formula above by finding $M$ different final call and put profit values.
For the M values I executed a data blocking ($M = 10000$ and $N_{blocks} = 100$), below are displayed the plots of global block averages.
The code written for these two functions is the following and can be found in `SOURCE/funzioni.h` (the data is in the folder `DATA`):
```cpp
// Black and Scholes analytical function for Call and Put
void Black_Scholes_Direct(vector<double> &C, vector<double> &P, Random rnd, int M)
{
   int S_0 = 100;         // asset_price_0
   double T_delivery = 1; // delivery_time
   double K = 100;        // Strike_price
   double r = 0.1;        // risk-free interest rate
   double sigma = 0.25;   // volatility

   // variables for calculations
   double S_t;
   double W;

   for (int i = 0; i < M; i++)
   {
      W = rnd.Gauss(0, T_delivery);

      S_t = S_0 * exp((r - pow(sigma, 2) / 2.0) * T_delivery + sigma * W); // Price at time

      // Call analytical
      C.push_back(exp(-r*T_delivery) * fmax(0, S_t - K));

      // Put analytical
      P.push_back(exp(-r*T_delivery) * fmax(0, K - S_t));
   }
   rnd.SaveSeed();
}

void Black_Scholes_Step_by_Step(vector<double> &C, vector<double> &P, Random rnd, int M)
{
   int S_0 = 100;         // asset_price_0
   double T_delivery = 1; // delivery_time
   double K = 100;        // Strike_price
   double r = 0.1;        // risk-free interest rate
   double sigma = 0.25;   // volatility

   // variables for calculations
   double S_t;
   double W;

   // size of step by step calculation
   const double h = 0.01;

   for (int i = 0; i < M; i++)
   {
   \\Function to write 1000 different Monte-Carlo step by step trajectories
      if (i <= 1000)
      {
         ofstream out;
         // Step by Step function
         S_t = S_0;
         string nomefile = "Traiettorie" + to_string(i);
         out.open("../DATA/" + nomefile + ".dat");
         for (int j = 0; j < 100; j++)
         {
            W = rnd.Gauss(0, T_delivery);
            S_t *= exp((r - pow(sigma, 2) / 2.0) * h + sigma * W * sqrt(h)); // Price at time
            if (j <= 50)
            {
               out << " " << j * h << "\t" << S_t << endl;
            }
         }
         out.close();
      }
      else
      {
         // Step by Step function
         S_t = S_0;
         for (int j = 0; j < 100; j++)
         {
            W = rnd.Gauss(0, T_delivery);
            S_t *= exp((r - pow(sigma, 2) / 2.0) * h + sigma * W * sqrt(h)); // Price at time
         }
      }

      // Call
      C.push_back(exp(-r) * fmax(0, S_t - K));

      // Put
      P.push_back(exp(-r) * fmax(0, K - S_t));
   }
   rnd.SaveSeed();
}
```

#### **Heatmap of Step-by-Step Trajectories**
Heatmap constructed from the first 1000 simulated trajectories obtained via the step-by-step discretization process.  
Each trajectory is generated independently and the corresponding data are stored in the `DATA` directory.

````python
import glob
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.cm as cm

# Prende tutti i file Traiettorie*.dat nella cartella data
files = sorted(glob.glob("data/Traiettorie*.dat"))
num_files = len(files)

# Impostazione stile coerente con i tuoi script precedenti[cite: 1, 2]
plt.style.use('seaborn-v0_8-whitegrid')
plt.figure(figsize=(10, 6))

# Genera un array di colori dalla colormap 'plasma'
colors = cm.plasma(np.linspace(0, 1, num_files))

# Cicla sui file e plotta ogni traiettoria singolarmente
for idx, file in enumerate(files):
    data = np.loadtxt(file)
    t = data[:, 0]
    S = data[:, 1]
    
    # Plot della singola traiettoria
    plt.plot(t, S, color=colors[idx], alpha=0.7, linewidth=1)

# Etichette e titolo
plt.xlabel("t", fontsize=12)
plt.ylabel("S(t)", fontsize=12)
plt.title("1000 Monte Carlo $S(t)$ trajectories", fontsize=14)

# Margini e visualizzazione
plt.tight_layout()
plt.savefig("IMAGES/traj.jpg")
plt.show()

````

This plot shows to demonstrate that the solution of a differential stochastic equation is in fact a hystogram of different possible trajectories. The precise stochastic differential equation for a geometric brownian motion is the following:
$$dS(t) = \mu S(t) dt + \sigma S(t) dW(t)$$
With the second component on the r.h.s which models the stochastic nature of the price evolution, which gives place to $Z_i \in \mathcal{N} (0,1)$ in the solution written above. It is possible to schematically write solutions to deterministic and stochastic differential and normal equations in the table below. 

|  | Deterministic | Stochastic |
| :--- | :--- | :--- |
| Classic Equation | $x$ (a scalar variable) | a hystogram |
| Differential equation | $x(t)$ (a trajectory along $t$) | a hystogram of trajectories|
