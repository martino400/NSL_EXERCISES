In Statistical Mechanics the one-dimensional (1D) Ising Model is exactly resoluble. The Hamiltonian for a system of N spins with nearest neighbors interaction is given by (in the following I will assume units such that $\mu_B=1$ and $k_B=1$):

$$
H = 
-J\sum_{i=1}^N s_i s_{i+1}
-\frac{h}{2}\sum_{i=1}^N (s_i + s_{i+1}) 
$$

where $\{ s_i \}_{i=1,N}$ are spin variables taking values in $\{-1;1\}$, $J>0$ (ferromagnetic model),and with the assumption $s_{N+1} = s_{1}$.

The partition function for $N$ spins, $Z$, can be exactly computed and it turns out to be:

$$
Z=\sum_{\{ s_i \}} \exp \left( -\beta H\right) = \lambda_1^N + \lambda_2^N
$$

where
$\lambda_{1,2} = 
e^{\beta J} 
\cosh (\beta h) \pm \left[ e^{2\beta J} \cosh^2 (\beta h) - 2\sinh (2\beta J) \right]^{1/2}$, 
and where $\beta=\frac{1}{k_b T} = \frac{1}{T}$.
Once the partition function has been computed, the Helmholtz free energy is readly obtained as $A = -k_B T \ln Z$. All the thermodynamic quantities of interest can be deduced from $A$ or, equivalently, from $Z$:

- the **Internal Energy** per spin (with $h=0$):

$$
\frac{U(N,T)}{N} = \frac{\langle H \rangle}{N}
= \frac{1}{N} \sum_{\{ s_i \}} H \frac{\exp \left( -\beta H\right)}{Z}
= -\frac{1}{N}\frac{\partial \ln Z}{\partial \beta}
\underset{h=0}= -J \tanh (\beta J) \Biggl( \frac{1 + \tanh^{N-2} (\beta J)}{1+\tanh^N (\beta J)} \Biggl)
$$

```python
import numpy as np
import matplotlib.pyplot as plt
points=100
T = np.linspace(0.2,3.0,num=points)
beta = 1/T
J = 1.0
Ns = 50
th = np.tanh(J/T)
thN= th**Ns
ch = 1/th
e = -J*( th + ch*thN )/( 1 + thN )
plt.plot(T, e)
plt.title('Ising 1D, internal energy')
plt.xlabel('T')
plt.ylabel('U/N')
plt.show()
```

- the **Heat Capacity** (with $h=0$):

$$
C(N,T) = \frac{\partial U(N,T)}{\partial T}
= \frac{\partial \beta}{\partial T}\frac{\partial U}{\partial \beta}
= -k_B\beta^2 \frac{\partial}{\partial \beta} \left[ \sum_{\{ s_i \}} H \frac{\exp \left( -\beta H\right)}{Z} \right] = \\
= -k_B\beta^2 \left[ -\sum_{\{ s_i \}} H^2 \frac{\exp \left( -\beta H\right)}{Z} + \sum_{\{ s_i \}} H \exp \left( -\beta H\right)\frac{\partial Z^{-1}}{\partial \beta} \right] = \\
= k_B\beta^2 \left\{ \sum_{\{ s_i \}} H^2 \frac{\exp \left( -\beta H\right)}{Z} + \left[ \sum_{\{ s_i \}} H \frac{\exp \left( -\beta H\right)}{Z}\right]\frac{1}{Z}\frac{\partial Z}{\partial \beta} \right\} =\\
= k_B\beta^2 \left\{ \sum_{\{ s_i \}} H^2 \frac{\exp \left( -\beta H\right)}{Z} - \left[ \sum_{\{ s_i \}} H \frac{\exp \left( -\beta H\right)}{Z}\right]^2 \right\} =
k_B\beta^2 \left(\langle H^2 \rangle -\langle H \rangle^2 \right)
$$

$$
\frac{C(N,T)}{N} = \frac{1}{N}\frac{\partial U(N,T)}{\partial T} =\\
\underset{h=0}= 
k_B (\beta J)^2 \left\{ 1+\tanh^N (\beta J)+(N-1)\tanh^2 (\beta J)+(N-1)\frac{\tanh^{N-2} (\beta J)}{1+\tanh^N (\beta J)}
- N \left[ \frac{\tanh (\beta J) + \tanh^{N-1} (\beta J)}{1+\tanh^N (\beta J)} \right]^2 \right\}
$$

```python
heat=((beta*J)**2)*(((1+thN+(Ns-1)*(th**2)+(Ns-1)*(ch**2)*thN)/(1+thN))-Ns*((th+ch*thN)/(1+thN))**2)
plt.plot(T, heat)
plt.title('Ising 1D, Heat capacity')
plt.xlabel('T')
plt.ylabel('C')
plt.show()

#Siccome sono in una dimensione non ho una transizione di fase definita, motivo per cui ho quella gobba
```

- The **magnetization** (with $h \not= 0$):

Note that with $h=0$ the magnetization $M(N,T,h=0)=\langle \sum_{i=1,N}s_i \rangle =0$ because in 1D we cannot have a (ferromagnetic) phase transition in absence of an external magnetic field.

$$
M(N,T,h) = -\frac{\partial A}{\partial h} = k_B T \frac{\partial \ln Z}{\partial h} =
\frac{1}{\beta} \frac{1}{Z}\frac{\partial}{\partial h}\sum_{\{ s_i \}} \exp \left( -\beta H\right) = \sum_{\{ s_i \}} \left[ \sum_{i=1,N}s_i\right] \frac{\exp \left( -\beta H\right)}{Z} = \left\langle \sum_{i=1,N}s_i \right\rangle
$$

The exact formula from $M(N,T,h) = k_B T \dfrac{\partial \ln Z}{\partial h}$ is:
$$
\frac{M(N,T,h)}{N}=\frac{e^{\beta J}\; \sinh(\beta h)}{Z} \; \Bigl[\lambda_1^{N-1}\cdot \Bigl(1+\frac{e^{\beta J}\; \cosh(\beta h)}{\sqrt{e^{2\beta J} \cosh^2 (\beta h) - 2\sinh (2\beta J)}}\Bigr) + \lambda_2^{N-1}\cdot \Bigl(1-\frac{e^{\beta J}\; \cosh(\beta h)}{\sqrt{e^{2\beta J} \cosh^2 (\beta h) - 2\sinh (2\beta J)}}\Bigr)\Bigr]
$$

```python
h=0.02 #external field
b = 1/T

l1 = np.exp(b*J)*np.cosh(b*h)+np.sqrt(np.exp(2*b*J)*np.cosh(b*h)*np.cosh(b*h)-2*np.sinh(2*b*J))
l2 = np.exp(b*J)*np.cosh(b*h)-np.sqrt(np.exp(2*b*J)*np.cosh(b*h)*np.cosh(b*h)-2*np.sinh(2*b*J))
Z = l1**Ns + l2**Ns
M = (np.exp(b*J)*np.sinh(b*h)*((l1**(Ns-1))*(1+np.exp(b*J)*np.cosh(b*h)/np.sqrt(np.exp(2*b*J)*np.cosh(b*h)*np.cosh(b*h)-2*np.sinh(2*b*J))) 
        + (l2**(Ns-1))*(1-np.exp(b*J)*np.cosh(b*h)/np.sqrt(np.exp(2*b*J)*np.cosh(b*h)*np.cosh(b*h)-2*np.sinh(2*b*J)))))/(Z)
plt.plot(T, M)
plt.title('Ising 1D, magnetization M with h = 0.02')
plt.xlabel('T')
plt.ylabel('$M$')
plt.show()

#Per Temperaturare sempre più basse ho sempre più magnetizzazione perchè ho sempre meno gli effetti di fluttuazione del sistema essendo il sistema "congelato", nei suoi gradi di libertà. 
```

- The **magnetic susceptibility** (with $h=0$):

$$
\chi(N,T) = \frac{\partial M}{\partial h} = 
\frac{\partial}{\partial h}\sum_{\{ s_i \}} \left[ \sum_{i=1,N}s_i\right] \frac{\exp \left( -\beta H\right)}{Z} =\\
= \beta\sum_{\{ s_i \}} \left[ \sum_{i=1,N}s_i\right]^2 \frac{\exp \left( -\beta H\right)}{Z}
-\sum_{\{ s_i \}} \left[ \sum_{i=1,N}s_i\right] \frac{\exp \left( -\beta H\right)}{Z}
\frac{1}{Z}\frac{\partial Z}{\partial h} = 
\beta \left[ \left\langle \left( \sum_{i=1,N}s_i \right)^2 \right\rangle - \left\langle \sum_{i=1,N}s_i \right\rangle^2 \right] = \\
\underset{h=0}= \beta \left[ \left\langle \left( \sum_{i=1,N}s_i \right)^2 \right\rangle \right] = \beta e^{2\beta J} \Biggl( \frac{1-\tanh^N (\beta J)}{1+\tanh^N (\beta J)} \Biggl)
$$

```python
X = beta*np.exp(2*beta*J)*(1-thN)/(1+thN)
plt.plot(T, X)
plt.title('Ising 1D, Susceptibility')
plt.xlabel('T')
plt.ylabel('$\chi$')
plt.show()

#Per temperatura molto basse la magnetizzazione del sistema risulta essere molto pronto alla variazione di campo esterno
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Set a clean, professional visual style
plt.style.use('seaborn-v0_8-whitegrid')

# 2. Data Loading
# Dati Gibbs
chi_gibbs = np.loadtxt('DATA/file0_gibbs.dat')
cv_gibbs = np.loadtxt('DATA/file2_gibbs.dat')
internal_gibbs = np.loadtxt('DATA/file1_gibbs.dat')
magn_gibbs = np.loadtxt('DATA/file3_gibbs.dat')

# Dati Metropolis
chi_metro = np.loadtxt('DATA/file0_metro.dat')
cv_metro = np.loadtxt('DATA/file2_metro.dat')
internal_metro = np.loadtxt('DATA/file1_metro.dat')
magn_metro = np.loadtxt('DATA/file3_metro.dat')

# 3. Create figure with 4 subplots (2x2 grid)
fig, axs = plt.subplots(2, 2, figsize=(14, 10))

# Palette colori per distinguere Gibbs e Metropolis
color_gibbs = '#E74C3C'  # Rosso per Gibbs
color_metro = '#2E86C1'  # Blu per Metropolis

# Helper function per plottare solo i punti con le barre d'errore (senza linee)
def plot_with_errors(ax, data, color, label, title, x_label=None, marker='o'):
    # x = col 0, y = col 2, yerr = col 4 (5th column)
    # fmt=marker disegna solo i punti. Es: marker='o' per cerchi, marker='s' per quadrati.
    ax.errorbar(data[:, 0], data[:, 2], yerr=data[:, 4], 
                color=color, fmt=marker, markersize=4, linewidth=1.2, 
                capsize=3, elinewidth=1, markeredgecolor='black', 
                alpha=0.8, label=label)
    
    ax.set_title(title, fontsize=14, fontweight='bold', pad=10)
    ax.set_ylabel('Value', fontsize=11, fontweight='bold')
    if x_label:
        ax.set_xlabel(x_label, fontsize=11, fontweight='bold')

# ==========================================
# [0, 0] Top Left: INTERNAL ENERGY
# ==========================================
# CORRETTO: axs[0, 0] per tutti
plot_with_errors(axs[0, 0], internal_gibbs, color_gibbs, 'Gibbs', 'Internal Energy (U)', marker='o')
plot_with_errors(axs[0, 0], internal_metro, color_metro, 'Metropolis', 'Internal Energy (U)', marker='s')
# THEORETICAL (Nota: T ed e devono essere definiti prima nel tuo script originale)
axs[0, 0].plot(T, e, color='black', linestyle='--', label='Theory', alpha=0.7)

# ==========================================
# [0, 1] Top Right: SPECIFIC HEAT
# ==========================================
# CORRETTO: axs[0, 1] per tutti
plot_with_errors(axs[0, 1], cv_gibbs, color_gibbs, 'Gibbs', 'Specific Heat ($C_v$)', marker='o')
plot_with_errors(axs[0, 1], cv_metro, color_metro, 'Metropolis', 'Specific Heat ($C_v$)', marker='s')
# THEORETICAL (Nota: T e heat devono essere definiti prima)
axs[0, 1].plot(T, heat, color='black', linestyle='--', label='Theory', alpha=0.7)

# ==========================================
# [1, 0] Bottom Left: MAGNETIZATION
# ==========================================
# CORRETTO: axs[1, 0] per tutti
plot_with_errors(axs[1, 0], magn_gibbs, color_gibbs, 'Gibbs', 'Magnetization (M)', x_label='Temperature (T)', marker='o')
plot_with_errors(axs[1, 0], magn_metro, color_metro, 'Metropolis', 'Magnetization (M)', x_label='Temperature (T)', marker='s')
# THEORETICAL (Nota: T e M devono essere definiti prima)
axs[1, 0].plot(T, M, color='black', linestyle='--', label='Theory', alpha=0.7)

# ==========================================
# [1, 1] Bottom Right: SUSCEPTIBILITY
# ==========================================
# CORRETTO: axs[1, 1] per tutti
plot_with_errors(axs[1, 1], chi_gibbs, color_gibbs, 'Gibbs', 'Susceptibility ($\chi$)', x_label='Temperature (T)', marker='o')
plot_with_errors(axs[1, 1], chi_metro, color_metro, 'Metropolis', 'Susceptibility ($\chi$)', x_label='Temperature (T)', marker='s')
# THEORETICAL (Nota: T e X devono essere definiti prima)
axs[1, 1].plot(T, X, color='black', linestyle='--', label='Theory', alpha=0.7)

# ==========================================
# Common Customization for all subplots
# ==========================================
for ax in axs.flat:
    # Remove top and right spines for a modern look
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    # Add legend with a slight shadow
    ax.legend(frameon=True, shadow=True, fontsize=10, loc='best')

# Global Title
fig.suptitle('Gibbs vs Metropolis Simulation Results', 
             fontsize=18, fontweight='bold', y=1.02)

# 4. Final layout optimization
plt.tight_layout()

# Save command (assicurati che la cartella 'Images' esista prima di lanciare lo script)
plt.savefig('Images/GibbsMetropolisPlot.png', dpi=300, bbox_inches='tight')

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Set a clean, professional visual style
plt.style.use('seaborn-v0_8-whitegrid')

# 2. Data Loading
# Dati Gibbs
chi_gibbs = np.loadtxt('DATA/file0_gibbs.dat')
cv_gibbs = np.loadtxt('DATA/file2_gibbs.dat')
internal_gibbs = np.loadtxt('DATA/file1_gibbs.dat')
magn_gibbs = np.loadtxt('DATA/file3_gibbs.dat')

# Dati Metropolis
chi_metro = np.loadtxt('DATA/file0_metro.dat')
cv_metro = np.loadtxt('DATA/file2_metro.dat')
internal_metro = np.loadtxt('DATA/file1_metro.dat')
magn_metro = np.loadtxt('DATA/file3_metro.dat')

# 3. Create figure with 4 subplots (2x2 grid)
fig, axs = plt.subplots(2, 2, figsize=(14, 10))

# Palette colori per distinguere Gibbs e Metropolis
color_gibbs = '#E74C3C'  # Rosso per Gibbs
color_metro = '#2E86C1'  # Blu per Metropolis

# Helper function per plottare i punti collegati da linee
def plots(ax, data, color, label, title, x_label=None, marker='o'):
    # x = col 0, y = col 4 (la quinta colonna degli errori)
    # linestyle='-' collega i punti con una linea continua
    ax.plot(data[:, 0], data[:, 4], 
            color=color, marker=marker, linestyle='-', linewidth=1.2, markersize=4, 
            markeredgecolor='black', alpha=0.8, label=label)
    
    ax.set_title(title, fontsize=14, fontweight='bold', pad=10)
    ax.set_ylabel('Value', fontsize=11, fontweight='bold')
    if x_label:
        ax.set_xlabel(x_label, fontsize=11, fontweight='bold')

# ==========================================
# [0, 0] Top Left: INTERNAL ENERGY
# ==========================================
plots(axs[0, 0], internal_gibbs, color_gibbs, 'Gibbs', 'Error of internal Energy (U)', marker='o')
plots(axs[0, 0], internal_metro, color_metro, 'Metropolis', 'Error of internal Energy (U)', marker='s')

# ==========================================
# [0, 1] Top Right: SPECIFIC HEAT
# ==========================================
plots(axs[0, 1], cv_gibbs, color_gibbs, 'Gibbs', 'Error of Specific Heat ($C_v$)', marker='o')
plots(axs[0, 1], cv_metro, color_metro, 'Metropolis', 'Error of Specific Heat ($C_v$)', marker='s')

# ==========================================
# [1, 0] Bottom Left: MAGNETIZATION
# ==========================================
plots(axs[1, 0], magn_gibbs, color_gibbs, 'Gibbs', 'Error of Magnetization (M)', x_label='Temperature (T)', marker='o')
plots(axs[1, 0], magn_metro, color_metro, 'Metropolis', 'Error of Magnetization (M)', x_label='Temperature (T)', marker='s')

# ==========================================
# [1, 1] Bottom Right: SUSCEPTIBILITY
# ==========================================
plots(axs[1, 1], chi_gibbs, color_gibbs, 'Gibbs', 'Error of Susceptibility ($\chi$)', x_label='Temperature (T)', marker='o')
plots(axs[1, 1], chi_metro, color_metro, 'Metropolis', 'Error of Susceptibility ($\chi$)', x_label='Temperature (T)', marker='s')

# ==========================================
# Common Customization for all subplots
# ==========================================
for ax in axs.flat:
    # Remove top and right spines for a modern look
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    # Add legend with a slight shadow
    ax.legend(frameon=True, shadow=True, fontsize=10, loc='best')

# Global Title
fig.suptitle('Gibbs vs Metropolis Error Results', 
             fontsize=18, fontweight='bold', y=1.02)

# 4. Final layout optimization
plt.tight_layout()

# Save command
plt.savefig('Images/GibbsMetropolisError.png', dpi=300, bbox_inches='tight')

plt.show()
```

- I did the simulation for a range of values of temperature from T = 2.0 to T = 0.5 by decreasing the temperature of 0.1 at every cycle. Furthermore, the for each point of the simulation I extrapolated the last value of the data-blocking done for each temperature, where for each temperature I used $N_{blocks} = 20 $ and $N_{steps} = 20000 $. 

- We can the that the data is compatible with the theoretical values. We can see that the Metropolis algorithm works better at lower temperatures.

- Another thing that it is possible to observe from the data obtained from both algorthsm: the errors on the data blocking become greater as the temperature decreases. In the Metropolis algorithm this is due to the fact that the probability of change becomes very small and therefore the number of Monte Carlo moves attempted becomes significantly greater than the number of MC moves accepted (acceptance becomes smaller, as plotted below). With Gibbs it doesn't make sense to talk about acceptance... (da capire bene cosa dire).

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Impostazioni di stile
plt.style.use('seaborn-v0_8-whitegrid')
plt.rcParams.update({
    'font.family': 'sans-serif',
    'font.size': 12,
    'axes.labelsize': 12,
    'axes.titlesize': 14
})

# Percorso del file
file_path = 'DATA/acceptanceMetro.dat'

# 2. Caricamento dei dati
# skiprows=1 salta l'intestazione testuale (TEMP: BLOCK TAKEN: ACCEPTED MC MOVES:)
# usecols=(0, 2) dice a numpy di leggere solo la prima colonna (0) e la terza (2)
data = np.loadtxt(file_path, skiprows=1, usecols=(0, 2))
    
# Separiamo le colonne (avendo estratto solo due colonne, i nuovi indici sono 0 e 1)
x = data[:, 0]  # Temperatura
y = data[:, 1]  # Accettazione Mosse MC
    
# 3. Creazione del grafico
plt.figure(figsize=(10, 6))
    
# Plottiamo i dati come una linea con i punti evidenziati (marker='o')
plt.plot(x, y, color='#1f77b4', marker='o', linewidth=2, markersize=6, label='$N_{accepted}$ / $N_{attempted}$')
    
# 4. Personalizzazione (Titolo, Etichette, Bordi)
plt.title('Metropolis acceptance w.r.t Temperature', fontweight='bold', pad=15)
plt.xlabel('Temperature (T)')
plt.ylabel('$N_{accepted}$ / $N_{attempted}$')
    
# Rimozione dei bordi superiori e destri per un look più pulito
ax = plt.gca()
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.spines['left'].set_color('#cccccc')
ax.spines['bottom'].set_color('#cccccc')
    
plt.legend(loc='best', frameon=True)
plt.tight_layout()
    
# 5. Salvataggio e visualizzazione
plt.savefig('Images/acceptance_plot.png', dpi=300, bbox_inches='tight')
plt.show()
```
